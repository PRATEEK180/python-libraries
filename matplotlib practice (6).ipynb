```python
companies=["BMW","Lamborghini","Bugatti"]
sales_2022=[2000,5000,12000]
sales_2023=[1000,6000,8000]
y=np.arange(len(companies))                #to assign index postition of the companies 
height=0.4                            #this is the thickness of each bar in the graph
plt.barh(y+0.2,sales_2022,height,label='2022')      #to shift the bars up and down for each company ,y depicts the index position of each company like 0=BMW,1=LAMBO,2=BUGA
plt.barh(y-0.2,sales_2023,height,label='2023')                 #Two bars for two years with respect to a single company so mentioned two times plt.barh()
plt.yticks(y,companies)                     #y describes position and companies say what to write on these positions.
plt.legend()
plt.show()

```

```python
x=[1,2,3,4,5]
y1=[10,20,30,40,50]
y2=[50,40,30,20,10]

plt.subplot(1,2,1)                            #1 row,2 column  and at position 1 , position 1 means on the layout you can adjust 
#position of the graphs where you want , if you give 3 at position , it will give error as you have only two columns
plt.plot(x,y1)
plt.title("graph 1")
plt.subplot(1,2,2)                   #subplot will make two different graphs on a single layout or figure 
plt.plot(x,y2)
plt.title("graph 2")

plt.show()
```

```python
x=[1,2,3,4]
y=[2,4,6,8]
z=[5,2,7,9]
fig,ax=plt.subplots(1,3)  #fig means one entire layout or figure, ax means the no of divisions on that single figure,
#this line says that on a single figure or layout , make 3 divisions
ax[0].plot(x,y) # and then on those 3 divisions , the first position ax[0] should be for the first plot(x,y) similarly for 2nd ,3rd plot
ax[0].set_title("graph1")
ax[1].plot(x,z)
ax[1].set_title("graph2")
ax[2].plot(y,z)
ax[2].set_title("graph3")
plt.show()
```

```python
ratings=[20,50,10,30]
players=["Tatum","lebron","brooks","castle"]
plt.pie(ratings,labels=players,autopct='%1.1f%%') #labels say what each slice represents ,using plt.pie(ratings)without the labels would just give you a simple pie chart with no labels 
plt.legend()   #autopct parameter for displaying the percentages of the pie in the chart
#%1.1f% mwans if percentageis 25 then it will be considered as 25.0% if you did %1.0f%% ,25 percent will be 25%
plt.show()
```

```python
marks=[99,50,87.4,94]
subjects=["maths","social","geography","science"]
c=["green","black","blue","yellow"]
plt.pie(marks,labels=subjects,radius=.5,colors=c) #like linewidth in line graph and markersize in scatter plot,, to adjust the size of the circle(pie chart , you have radius
plt.legend() #using color argument to use the desired color for visualization of the pie chart
plt.show()
```

```python
exam_scores=[85,92,78,90,88,95,82,79,87,91]
def calculate_exam_statistics(exam_scores):
    num_students=len(exam_scores)
    average_score=sum(exam_scores)/num_students
    highest_score=max(exam_scores)
    lowest_score=min(exam_scores)
    num_passed=0
    for score in exam_scores:
        if score>=80:
            num_passed= num_passed+1
    num_failed=num_students-num_passed
    return average_score,highest_score,lowest_score,num_passed,num_failed

       
```

```python
exam_scores=[85,92,78,90,88,95,82,79,87,91]
average_score,highest_score,lowest_score,num_passed,num_failed= calculate_exam_statistics(exam_scores)
print("the average score is",average_score)
print("the highest is",highest_score)
print("the lowest score is",lowest_score)
print("total students passed", num_passed)
print("total students failed",num_failed)
```

```python
def pass_valid(password):
    has_letter=False
    has_number=False
    has_special=False
    special_char="!@#$%^&*"
    for letter in password:
        if letter.isalpha():
            has_letter=True
        elif letter.isdigit():
            has_number=True
        elif letter in special_char:
            has_special=True
    if has_letter and has_number and has_special:
        return "password is valid"
    else:
        return "not valid"
        

```

```python
def sum_unique_elements(numbers):
    unique_elements=[]
    total=0
    for number in numbers:
        if number not in unique_elements:
            unique_elements.append(number)
            total=total+number
    return total
    
```

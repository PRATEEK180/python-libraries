1. Write a python function `is_alternating_list()` that checks if a python list satisfies the criterion of alternating even and odd values throughout. The function will return two values:
* True or False depending on whether the list satisfies the above criterion.
* The index in which the alternating sequence is broken (this value will be -1 as the index when True and hence sequence is never broken).

For example:
* For the list [7, 2, 5, 4, 3, 16, 9], the function should return True and -1.
* For the list [7, 2, 5, 4, 3, 16, 10], the function should return False and 6 (index of value 10 which breaks the sequence).
* For the list [7, 2, 0, 4, 3, 16, 9], the function should return False and 2 (index of value 0 which breaks the sequence)

```python
# write your code here


def is_alternating_list(l):
    for i in range(len(l)-1):  # l-1 because cant go to extreme end of the list as the index position will always be one less than the range
        if l[i]%2==l[i+1]%2:   #the approach is to compare pair values side-by-side , if both even or both odd , then thats a False 
            return False,i+1   # since we are checking the values in pairs, its always the second no of the pair thats breaking the pattern,else the first no was fine  by itself as it was obeying the pattern
    return True,-1            # as mentioned in the question that index for True will be -1
                
```

2. Write a function `longest_sequence_by_parity()` which, given a list of integer numbers as input, identifies and outputs the longest sequence of even or odd numbers that can be found within the list.

For example:
* `longest_sequence_by_parity([6, 1, 3, 5, 7, 2, 4, 8, 7, 1, 2])` will output [1, 3, 5, 7] as this is the longest sequence that retains the same parity, in this case a sequence of 4 odd values
* `longest_sequence_by_parity([1, 7, 2, 4, 6, 8, 2, 2, 3, 5, 5, 5])`will output [2, 4, 6, 8, 2, 2] as this is the longest sequence that retains the same parity, and in this case it is a sequence of 6 even values.

If there is a tie (i.e. two sequences within the list are equally long), it will output the first occurrence starting from the left.

```python
def longest_sequence_by_parity(l):
    longest_sequence=[]
    current_sequence=[]
    for numbers in l:
        if len(current_sequence)==0:
            current_sequence.append(numbers)
        elif numbers%2==current_sequence[-1]%2:
            current_sequence.append(numbers)
        else:
            if len(current_sequence)>len(longest_sequence):
                longest_sequence=current_sequence

            current_sequence=[numbers]
    if len(current_sequence)>len(longest_sequence):
        longest_sequence=current_sequence
    return longest_sequence
                
                
                
```

```python

longest_sequence_by_parity([1, 1, 2, 2, 2, 10, 14, 26, 3, 3, 3, 5, 4, 4, 3, 7, 2, 2])
```

Write a function `is_symmetric()` which determines if a matrix given as input is symmetric both horizontally and vertically, and the number of values that need to be amended for it to become symmetric.

For example, matrix1 is symmetric, so it would return True and 0 (no amendments needed); matrix2 is not symmetric, and in this case it would return False and 2 (i.e. there are two values, either value from each pair, that need amending for it to become symmetric, as highlighted in red).

The matrix can have any size, `n x m`.

```python
# write your code here
def is_symmetric(matrix):
    amendments=0
    rows=len(matrix)
    columns=len(matrix[0])
    for row in range(rows//2):   #for horizontal symmetry
        opposite_row=rows-1-row
        for column in range(columns):
            if matrix[row][column]!= matrix[opposite_row][column]:
                amendments=amendments+1
    for row in range(rows):         #for vertical symmetry
        for column in range(columns//2): 
            opposite_column=columns-1-column
            if matrix[row][column]!= matrix[row][opposite_column]:
                amendments=amendments+1

    if amendments==0:
        return True,0
    else:
        return False, amendments
                
        
```

5. For this exercise, we will only consider users who have rated at least 10 movies. We will look at negativity and positivity ratios to identify the most negative and the most positive users:
* Positivity ratio: the fraction of ratings that a user has rated between 3.0 and 5.0 (both inclusive). If a user has rated 20 movies, and 12 of them are within the range, their positivity ratio = 12 / 20 = 0.6
* Negativity ratio: the fraction of ratings that a user has rated between 0.5 and 2.5 (both inclusive). If a user has rated 20 movies, and 6 of them are within the range, their negativity ratio = 6 / 20 = 0.3

Write a function `most_positive_and_negative()` which outputs the full name (name + last name) of the user with the highest positivity ratio and the full name of the user with the highest negativity ratio.

In case of a tie, return the first of the instances.

```python
# write your code here
import csv
def most_positive_and_negative():
    total_ratings={}
    positive_ratings={}
    negative_ratings={}
    with open("ratings.csv","r") as file:
        reader=csv.DictReader(file)

        for row in reader:
            user_id=row["userId"]
            rating=float(row["rating"])

            if user_id not in total_ratings:
                total_ratings[user_id]=0
                positive_ratings[user_id]=0
                negative_ratings[user_id]=0


            total_ratings[user_id]=total_ratings[user_id]+1

            if rating>=3 and rating<=5:
                positive_ratings[user_id]=positive_ratings[user_id]+1
            if rating>=0.5 and rating<=2.5:
                negative_ratings[user_id]=negative_ratings[user_id]+1

    most_positive_user= None
    most_negative_user=None
    highest_positive_ratio=-1
    highest_negative_ratio=-1
    for user_id in total_ratings:
        if total_ratings[user_id]>=10:

            positive_ratio=positive_ratings[user_id]/ total_ratings[user_id]
            negative_ratio=negative_ratings[user_id]/total_ratings[user_id]

            if positive_ratio>highest_positive_ratio:
                highest_positive_ratio=positive_ratio
                most_positive_user=user_id
            if negative_ratio>highest_negative_ratio:
                highest_negative_ratio=negative_ratio
                most_negative_user=user_id

    users={}
    with open("users.csv","r") as file:
        reader=csv.DictReader(file)
        for row in reader:
            user_id=row["userId"]
            name=row["name"]
            last_name=row["lastname"]

            users[user_id]=name + " " + last_name

    return users[most_positive_user],users[most_negative_user]
```

The following cell will run the function and produce the output:

```python
most_positive_and_negative()
```

6. We are looking for the title of the top-rated movie by a set of at least 5 users selected based on their education.

A question we could ask is, for example, what is the top-rated movie by users from the 'Master' group? Or what is the top-rated movie by users from the 'Master' and 'Doctoral' groups combined?

This is what we want to achieve:
1. Write a function `top_rated_by_education()` which takes a list of education values as input, which can be a single element or multiple, e.g. ['Master'], or ['Master', 'Doctoral'].
2. We will rely on ratings given only by users who belong to any of the education groups provided.
3. Looking at the ratings of users pertaining to the selected groups only, and only relying on movies rated by 5 or more of these users, what's the title of the movie with the highest average rating?

```python
import csv 
import json
def top_rated_by_education(education_values):
    selected_users={}
    with open("users.csv","r")as file:
        reader=csv.DictReader(file)
        for row in reader:
            if row["education"]in education_values:
                selected_users[row["userId"]]=True

    total_rating={}
    number_of_ratings={}
    with open("ratings.csv","r") as file:
        reader=csv.DictReader(file)
        for row in reader:
            if row["userId"] in selected_users:
                movie_id=row["movieId"]
                rating=float(row["rating"])
                if movie_id not in total_rating:
                    total_rating[movie_id]=0
                    number_of_ratings[movie_id]=0
                total_rating[movie_id]=total_rating[movie_id]+rating
                number_of_ratings[movie_id]+=1
    best_movie=None
    highest_average=-1
    for movie_id in total_rating:
        if number_of_ratings[movie_id]>=5:
            average=total_rating[movie_id]/number_of_ratings[movie_id]
            if average>highest_average:
                highest_average=average
                best_movie=movie_id
    with open("movies.json","r") as file:
        for line in file:
            movie=json.loads(line)
            if movie["id"]==best_movie:
                return movie["title"]
    
```

As an example, produce the output for the combined group of users including 'Master' and 'Doctoral' education.

```python
top_rated_by_education(['Master', 'Doctoral'])
```

```python
import pandas as pd
```

```python
df.groupby("address")["G3"].max() # which address type (urban/rural) have higher grades {urban has higher grades(20)}
```

```python
df[df["G3"]>15].shape[0]  # no of students who have grades higher than 15
```

```python
failed=df[df["G3"]<10] # STUDENTS WHO FAILED, THAT IS GRADES BELOW 10
```

```python
print(failed.shape[0])
```

```python
print(failed[["sex","age","G3"]]) #filter the failed students based on age sex and grades 
```

```python
# Dalc = workday alcohol, Walc = weekend alcohol
print(df.groupby("Dalc")["G3"].mean())  # does alcohol consumption affect grades?
```

```python
print(df['internet'])
```

```python
print(df.groupby("internet")["G3"].mean()) #has internet usage affected the students's grades? no-9.4 and yes-10.6, means using internet has helped students score more marks. 
```

```python
print(df.corr(numeric_only=True)["G3"].sort_values(ascending=False))#which factors affecting the grades the most, Failures affect the grades
```

```python
 #how many students have improved from G1 to G3? 
df['improvement']=df["G3"]-df["G1"] #make new column "improvement"
```

```python
improved=df[df["improvement"]>0] #if the difference in improvement is greater than 0 it is considered as improved
```

```python
print(improved.shape[0]) #143 students have improved from G1 to G3
```

```python
print(df.groupby("sex")["Walc"].mean()) #females drink less alcohol(1.9) compared to males (2.6)
```

```python
# IMPORTANT FINDINGS
# 1. Higher alcohol consumption = lower grades!
# 2. Urban students perform better than rural!
# 3. More study time = better grades!
# 4. Internet access positively affects grades!
# 5. G2 is the strongest predictor of G3!
# 6. Males consume more alcohol than females!
# 7. 130 students failed (grade below 10)!
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline
```

```python
raw_data = {'first_name': ['Jason', 'Molly', 'Tina', 'Jake', 'Amy'], 
            'last_name': ['Miller', 'Jacobson', 'Ali', 'Milner', 'Cooze'], 
            'female': [0, 1, 1, 0, 1],
            'age': [42, 52, 36, 24, 73], 
            'preTestScore': [4, 24, 31, 2, 3],
            'postTestScore': [25, 94, 57, 62, 70]}

df = pd.DataFrame(raw_data, columns = ['first_name', 'last_name', 'age', 'female', 'preTestScore', 'postTestScore'])

df
```

```python
plt.scatter(df.preTestScore, df.postTestScore, s=df.age)

#set labels and titles
plt.title("preTestScore x postTestScore")
plt.xlabel('preTestScore')
plt.ylabel('preTestScore')
plt.show()
```

```python
plt.scatter(df.preTestScore, df.postTestScore, s= df.postTestScore * 4.5, c = df.female)

#set labels and titles
plt.title("preTestScore x postTestScore")
plt.xlabel('preTestScore')
plt.ylabel('preTestScore')
plt.show()
```
